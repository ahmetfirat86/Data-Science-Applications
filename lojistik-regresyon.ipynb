```python
# UCI Machine Learning Repository'den veri seti indirmeye yarayan fonksiyonu içe aktarır.
from ucimlrepo import fetch_ucirepo
# Veriyi eğitim ve test kümelerine ayıran fonksiyonu içe aktarır.
from sklearn.model_selection import train_test_split
# Lojistik regresyon sınıflandırma modelini içe aktarır.
from sklearn.linear_model import LogisticRegression
# Tablo biçiminde veri işlemek için pandas kütüphanesini içe aktarır.
import pandas as pd
```

```python
# Uyarı mesajlarını yönetmek için standart warnings modülünü içe aktarır.
import warnings
# Bu betik çalışırken uyarı mesajlarının ekrana yazılmamasını sağlar.
warnings.filterwarnings("ignore")
```

```python

# UCI deposundan "heart_disease" adlı veri setini indirir/yükler.
heart_disease = fetch_ucirepo(id=45)
```

```python
# Veri setindeki girdi özelliklerini bir pandas DataFrame tablosuna dönüştürür.
df = pd.DataFrame(data = heart_disease.data.features)
```

```python
# Hedef sınıf değerlerini "target" isimli sütun olarak tabloya ekler.
df["target"] = heart_disease.data.targets
```

```python
# Tabloda herhangi bir eksik değer bulunup bulunmadığını kontrol eder.
if df.isna().any().any():
    # Eksik değeri bulunan satırları tablodan yerinde siler.
    df.dropna(inplace = True)
    # Eksik değer bulunduğunu belirten mesajı ekrana yazdırır.
    print("nan")
```

```python
# Hedef sütunu dışındaki tüm sütunları modelin girdi özellikleri olarak alır ve NumPy dizisine dönüştürür.
X = df.drop(["target"], axis = 1).values
# "target" sütunundaki sınıf etiketlerini NumPy dizisi olarak alır.
y = df.target.values
```

```python
# Veriyi yüzde 90 eğitim ve yüzde 10 test kümelerine ayırır; sabit tohum bölmeyi tekrarlanabilir yapar.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.1, random_state = 42)
```

```python
# L2 düzenlileştirmeli, lbfgs çözücülü lojistik regresyon modelini en fazla 100 iterasyonla oluşturur.
log_reg = LogisticRegression(penalty="l2", C=1, solver="lbfgs", max_iter=100)
# Modeli eğitim özellikleri ve bunlara karşılık gelen sınıf etiketleriyle eğitir.
log_reg.fit(X_train, y_train)
```

```python
# L2 düzenlileştirmeli, lbfgs çözücülü lojistik regresyon modelini en fazla 100 iterasyonla oluşturur.
log_reg = LogisticRegression(penalty="l2", C=1, solver="lbfgs", max_iter=100)
# Modeli eğitim özellikleri ve bunlara karşılık gelen sınıf etiketleriyle eğitir.
log_reg.fit(X_train, y_train)
```

```python
# Eğitilmiş modelin test kümesindeki doğruluk oranını hesaplar.
accuracy = log_reg.score(X_test, y_test)
# Test doğruluğunu ekrana yazdırır.
print("Logistic Regression Acc:", accuracy)



```
